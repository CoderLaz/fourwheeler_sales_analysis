```python
import pandas as pd
import ast

df = pd.read_csv('sales_raw.csv', sep='|')
df['sales'] = df['sales'].apply(lambda x: ast.literal_eval(x))

df = pd.concat([df, df['sales'].apply(pd.Series)], axis=1)
df.drop(columns=['sales'], inplace=True)

df.dropna(subset=df.columns.drop(['make', 'model']), inplace=True)

df = df.melt(id_vars=['make', 'model'], var_name='date', value_name='sales')
df['month'] = df['date'].apply(lambda x: x.split(' ')[0])
df['year'] = df['date'].apply(lambda x: '20' + x.split(' ')[1])
df = df[['make', 'model', 'date', 'month', 'year', 'sales']]
df.to_csv('sales.csv', index=False)
df
```

## Aim
As part of this data analysis project, the goal of this notebook is to organize the raw data through pandas so it gets simple to analyze and plot the visualization.

## Procedure
1. Parse the sales cell as dict
2. Convert the dicts to Series so they each Month-Year act as separate columns
3. Drop the rows which contain NaN sales values
4. Transpose the Month-Year columns (like Jan 20, Feb 20, Mar 20, ...) into a two columns one containing Month-Year and another corresponding Sales figures
5. Visualization & Analysis

Requirements: The packages installed can be found in the requirements.txt basically I have used pandas and most other packages like ipykernel, etc are installed due to running jupyter notebook in vs code

1. Parse the sales cell as dict

```python
import pandas as pd
import ast

'''
Here we just read the csv and then using apply method in pandas which can be used to modify a cell in dataframe.
We can just pass the function in the apply as well, lambda functions are one line functions which can be used for single line tasks to get things done in one go.
ast.literal_eval is a built in function which is used to convert string to dict
Uncommenting below comments, it can be checked the datatype of previous sales values and modified sales values
'''
df = pd.read_csv('sales_raw.csv', sep='|')
# print(type(df['sales'][0]))
df['sales'] = df['sales'].apply(lambda x: ast.literal_eval(x))
# print(type(df['sales'][0]))
```

2. Convert the dicts to Series so they each Month-Year act as separate columns

```python
'''
Here pd.Series just ellaborates each dict containing Month-Year and sales values into each column of Month-Year containing their values
inplace=True can be used to modify the original df without reassigning it
'''

df = pd.concat([df, df['sales'].apply(pd.Series)], axis=1)
df.drop(columns=['sales'], inplace=True)
df

```

3. Drop the rows which contain NaN sales values

```python
'''
subset is used to consider the columns for dropping NaN values, basically to drop the row if certain columns have NaN values
'''

df.dropna(subset=df.columns.drop(['make', 'model']), inplace=True)
# df.shape
```

4. Transpose the Month-Year columns (like Jan 20, Feb 20, Mar 20, ...) into a two columns one containing Month-Year and another corresponding Sales figures

```python
'''
melt is used for the same operation where we just need transpose Month-Year columns as two columns one having Month-Year name and the other it's corresponding sales.
Also these are represented by var_name and value_name params as below.
As mentioned in step 1 we use apply function to just extract months and years separately and save them.
The resultant dataframe is saved as sales.csv.

The next step is to plot the visualization and analyze the data.
'''

df = df.melt(id_vars=['make', 'model'], var_name='date', value_name='sales')
df['month'] = df['date'].apply(lambda x: x.split(' ')[0])
df['year'] = df['date'].apply(lambda x: '20' + x.split(' ')[1])
df = df[['make', 'model', 'date', 'month', 'year', 'sales']]
df.to_csv('sales.csv', index=False)
df
```
